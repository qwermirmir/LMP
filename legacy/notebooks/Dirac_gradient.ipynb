```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
using LinearAlgebra
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 0
Dim = 4
NC = 3
# const pes =  (2, 2, 2, 2)
# const cuda = true
# const numtrj_in = 2 # the number of operations to take average of performance
# const MDsteps_in = 0 # dummy




U = Initialize_Gaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))
```

```julia
params = Dict()
params["Dirac_operator"] = "Wilson"
params["κ"] = 0.141139
params["eps_CG"] = 1.0e-8
params["verbose_level"] = 2

D = Dirac_operator(U,x,params)
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
W=Generate_weight(hop)
W1=deepcopy(W)
W2=Generate_weight(hop)
```

```julia
layer1=CPTC_Layer(hop,W)
layer2=CPTC_Layer(hop,W1)
# layer3=CPTC_Layer(hop,W2)
layer3=Dirac_Layer(D)
```

```julia
layers=[layer2,layer3]
y=layer1(x)
```

```julia
delta=1e-9
```

```julia
f=norm2(layers(y)-t)
grad=deepcopy(W)
```

```julia
epsilon = 1e-6


for i in 1:9
    for u in 1:4
        for v in 1:4
            cW=deepcopy(W)
            cW[i][u,v]=cW[i][u,v]+epsilon
            layer1=CPTC_Layer(hop,cW)
            y1=layer1(x)
            grad[i][u,v]=(norm2(layers(y1)-t)-f)/epsilon
        end
    end
end 
```

```julia
layers=[layer1,layer2,layer3]
γ=Clifford()
u = forward_propagation(layers, x,t)
delta = backward_propagation(layers, t, u)
d = gradient(layer1, layer1.activation, u[1], delta[3],γ)
```
