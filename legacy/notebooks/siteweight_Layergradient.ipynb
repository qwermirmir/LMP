```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 0
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))





delta=1e-8
P=Onestep()
hop=Hopping_operator(U,P)
W=Generate_weight(hop)




```

```julia
layer=PTC_Layer(hop,W)
act=Identity()
Attention=siteweight_Layer(layer,0.1, act ,false)
layers=[Attention]
y=layers(x)
f=norm2(y-t)
grad1=deepcopy(Attention.Q.W)
grad2=deepcopy(Attention.K.W)
grad3=deepcopy(Attention.V.W)
```

```julia
norm2(y)
```

```julia

epsilon=1e-8


for i in 1:length(Attention.Q.W)
    for u in 1:4
        for v in 1:4
            local layer1=deepcopy(layers)
            layer1[1].Q.W[i][u,v]+=epsilon
            y1=layer1(x)
            grad1[i][u,v]=(norm2(y1-t)-f)/epsilon
        end
    end
end 

for i in 1:length(Attention.K.W)
    for u in 1:4
        for v in 1:4
            local layer1=deepcopy(layers)
            layer1[1].K.W[i][u,v]+=epsilon
            y1=layer1(x)
            grad2[i][u,v]=(norm2(y1-t)-f)/epsilon
        end
    end
end 
for i in 1:length(Attention.V.W)
    for u in 1:4
        for v in 1:4
            local layer1=deepcopy(layers)
            layer1[1].V.W[i][u,v]+=epsilon
            y1=layer1(x)
            grad3[i][u,v]=(norm2(y1-t)-f)/epsilon
        end
    end
end
```

```julia
γ=Clifford()
u = forward_propagation(layers, x,t)
delta = backward_propagation(layers, t, u)

d = gradient(layers[1], layers[1].activation, u[1], delta[1],γ)
```

```julia
println(grad1[1]-d[1][1])
println(grad2[2]-d[2][2])
println(grad3[3]-d[3][3])
```
