```julia
f = spin_color_vector(U) 
f[2][1,1,1,1,1,1]
```

```julia
# Spin_matrix 및 spin_color_vector 초기화
W = Random_SpinM("complex")
W_prime = Random_SpinM("complex")
 # U는 Initialize_pseudofermion_fields 등을 통해 생성

# 연산 순서를 바꿔 비교
# lhs =  PTC([W_prime], Hopping_operator(U, Paths([1, 2, 3])), f)
# rhs = PTC([W_prime], Hopping_operator(U, Paths([1, 2, 3])), f)

lhs = W * PTC([W_prime], Hopping_operator(U, Paths([1, 2, 3])), f)
rhs = PTC([W_prime], Hopping_operator(U, Paths([1, 2, 3])), W * f)

# 결과 비교
println("LHS: ", lhs)
println("RHS: ", rhs)

```

```julia
sub=lhs-rhs
if inner_product(sub,sub)==0
    println("교환 가능!")
else
    println("교환 불가능!")
end
```
