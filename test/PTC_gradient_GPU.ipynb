```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
using LinearAlgebra
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 0
Dim = 4
NC = 3
const pes =  (2, 2, 2, 2)
const cuda = true
const numtrj_in = 2 # the number of operations to take average of performance
const MDsteps_in = 0 # dummy
PEs = pes

# U=Initialize_Gaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
U = Initialize_Gaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot",cuda=true,blocks=PEs)
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))
```

```julia
params = Dict()
params["Dirac_operator"] = "Wilson"
params["κ"] = 0.2563
params["faster version"] = true
params["eps_CG"] = 1.0e-8
params["verbose_level"] = 3

D = Dirac_operator(U,x,params)
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
W=Generate_weight(hop)
W1=Generate_weight(hop)
W2=Generate_weight(hop)
```

```julia
layer1=PTC_Layer(hop,W)
layer2=PTC_Layer(hop,W1)
layer3=PTC_Layer(hop,W2)
```

```julia
layers=[layer2,layer3]
y=layer1(x)
```

```julia
delta=1e-9
```

```julia
f=norm2(layers(y)-t)
grad=deepcopy(W)
```

```julia
println(f)
```

```julia
epsilon = 1e-5


for i in 1:9
    for u in 1:4
        for v in 1:4
            cW=deepcopy(W)
            cW[i][u,v]=cW[i][u,v]+epsilon
            layer1=PTC_Layer(hop,cW)
            y1=layer1(x)
            grad[i][u,v]=(norm2(layers(y1)-t)-f)/epsilon
        end
    end
end 
```
