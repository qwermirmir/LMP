```julia
using LatticeDiracOperators
using LatticeQCD
using Wilsonloop
using Gaugefields
include("elementry_base.jl")
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 0
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
```

```julia
gauss_distribution_fermion!(x)
```

```julia
y=similar(x)
```

```julia
for i in 1:4
    println("x[1,$i,1,1,1,1] = ",x[1,i,1,1,1,1])
end

shift_x = shift_fermion(x, -1)
for i in 1:4
    println("shift_x[1,$i,1,1,1,1] = ", shift_x[1,i,1,1,1,1])
end
```

```julia
println("before")
for i in 1:4
    println("y[1,$i,1,1,1,1] = ", y[1,i,1,1,1,1])
end
mul!(y, U[1], shift_x)
println("after")
for i in 1:4
    println("y[1,$i,1,1,1,1] = ", y[1,i,1,1,1,1])
end
```

```julia
println("before")
for i in 1:4
    println("y[1,$i,1,1,1,1] = ", y[1,i,1,1,1,1])
end
mul!(y, U[1]', y)
println("after")
for i in 1:4
    println("y[1,$i,1,1,1,1] = ", y[1,i,1,1,1,1])
end
```

```julia
shift_y = shift_fermion(y, 1)
for i in 1:4
    println("shift_y[1,$i,1,1,1,1] = ", shift_y[1,i,1,1,1,1])
end
```
