```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 1
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
```

```julia
W=Generate_weight(hop)
```

```julia
layer1 = CPTC_Layer(hop, W)
```

```julia
G=Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
```

```julia
layer2=CPTC_Layer(shop, W)
```
