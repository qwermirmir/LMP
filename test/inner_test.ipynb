```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 1
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
```

```julia
gamma = Clifford()
W=Generate_weight(hop)
```

```julia
layer1 = CPTC_Layer(hop, W)
```

```julia
s = hop(x)
grad = deepcopy(W)
```

```julia
W = layer1.W
lW = length(W)
```

```julia
result, u = forward_propagation([layer1], x)
```

```julia
f = (u[end] - t)

    
d1= conj(f)/2

```

```julia

cW = conj(Clifford(layer1.W))
invP = reverse_paths(hop.P)
invhop = Hopping_operator(hop.Gauge, invP)

cd=conj(d1)
cL=CPTC_Layer(invhop,cW)

result=cL(cd)
result=conj(result)
```

```julia
for i in 1:lW
    k = []
    for j in 1:16
        b = gamma[j] * s[i]
        push!(k, inner_product(conj(result), b))
    end
    for u in 1:4
        for v in 1:4
            grad[i][u, v] = k[(u-1) * 4 + v]
        end
    end
end 
grad=grad+conj(grad)
```

```julia
subgrad=deepcopy(W)

for i in 1:9



    for u in 1:4
        for v in 1:4
            W2=deepcopy(W)
            W3=deepcopy(W)
            delta = 1e-9
            W2[i][u, v] = W2[i][u, v]+delta
            W3[i][u, v] = W3[i][u, v]-delta
            layer2 = CPTC_Layer(hop, W2)
            layer3 = CPTC_Layer(hop, W3)
            L=(Lossfunction(layer2(x),t)-Lossfunction(layer3(x),t))/(2*delta)

            subgrad[i][u, v]=L



        end
    end

end 
```
