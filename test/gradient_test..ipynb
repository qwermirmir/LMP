```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 1
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
```

```julia
W=Generate_weight(hop)
```

```julia
subgrad=deepcopy(W)

for i in 1:9



    for u in 1:4
        for v in 1:4
            W2=deepcopy(W)
            W3=deepcopy(W)
            delta = 1e-9
            W2[i][u, v] = W2[i][u, v]+delta
            W3[i][u, v] = W3[i][u, v]-delta
            layer2 = CPTC_Layer(hop, W2)
            layer3 = CPTC_Layer(hop, W3)
            L=(Lossfunction(layer2(x),t)-Lossfunction(layer3(x),t))/(2*delta)

            subgrad[i][u, v]=L



        end
    end

end 


```
