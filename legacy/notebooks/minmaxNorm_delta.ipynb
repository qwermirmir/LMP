```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
using LinearAlgebra
```

```julia
NX = 1
NY = 1
NZ = 1
NT = 1
Nwing = 0
Dim = 16
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
W1=Generate_weight(hop)
W2=Generate_weight(hop)
```

```julia
layer1=CPTC_Layer(hop,W1)
# layer2=CPTC_Layer(hop,W2)
act=Softmax()

layer1=Attention_Layer(layer1,sqrt(length(x)*1000)  ,act , true )
layer2=minmaxNormLayer()
layers=[layer1,layer2]
```

```julia
y=layer1(x)
```

```julia
f=norm2(layer2(y)-t)
grad1=deepcopy(layer1.Q.W)
grad2=deepcopy(layer1.K.W)
grad3=deepcopy(layer1.V.W)
```

```julia
println(f)
```

```julia
γ=Clifford()
u = forward_propagation(layers, x)
delta = backward_propagation(layers, t, u)

d = gradient(layers[1], layers[1].activation, u[1], delta[2],γ)
```

```julia

epsilon=1e-8

for i in 1:9
    for u in 1:4
        for v in 1:4
            local layer3=deepcopy(layers)
            layer3[1].Q.W[i][u,v]+=epsilon
            y1=layer3[1](x)
            grad1[i][u,v]=(norm2(layer3[2](y1)-t)-f)/epsilon
        end
    end
end 

for i in 1:9
    for u in 1:4
        for v in 1:4
            local layer3=deepcopy(layers)
            layer3[1].K.W[i][u,v]+=epsilon
            y1=layer3[1](x)
            grad2[i][u,v]=(norm2(layer3[2](y1)-t)-f)/epsilon
        end
    end
end

for i in 1:9
    for u in 1:4
        for v in 1:4
            local layer3=deepcopy(layers)
            layer3[1].V.W[i][u,v]+=epsilon
            y1=layer3[1](x)
            grad3[i][u,v]=(norm2(layer3[2](y1)-t)-f)/epsilon
        end
    end
end 
```
