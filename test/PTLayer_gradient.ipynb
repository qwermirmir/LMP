```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
using LinearAlgebra
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 0
NC = 3

U = Initialize_4DGaugefields(NC, Nwing, NX, NY, NZ, NT, condition = "hot")
x = Initialize_pseudofermion_fields(U[1], "Wilson", nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1], "Wilson", nowing=true)
gauss_distribution_fermion!(t)
x = x / sqrt(norm2(x))
t = t / sqrt(norm2(t))
```

```julia
params = Dict()
params["Dirac_operator"] = "Wilson"
params["κ"] = 0.14
params["faster version"] = false   # Wilson_Dirac_operator_faster는 * 메서드 미정의
params["eps_CG"] = 1.0e-8
params["verbose_level"] = 2

D = Dirac_operator(U, x, params)
```

```julia
# 9방향 one-step paths: identity + ±1,±2,±3,±4
npaths = 9
P = Vector{Paths}(undef, npaths)
for i in 1:npaths
    if i <= 5
        P[i] = Paths([[i-1]])
    else
        P[i] = Paths([[-(i-5)]])
    end
end
```

```julia
n_out = 1
PT   = PTLayer(P, U, n_out)
PT.W = 0.1 * PT.W

fe   = fermionfield_Layer(x, npaths)
loss = PtcRayleighQuotient_Layer(D)

layers = [fe, PT, loss]
```

```julia
# forward pass — f: loss 값
u, f = forward_propagation(layers, D, D)
println("loss f = ", f)
```

```julia
# 수치 gradient (실수부): W[k][u,v] → +epsilon 섭동
epsilon = 1e-6
NW = length(PT.W)
num_grad_re = deepcopy(PT.W)

for k in 1:NW
    for uu in 1:4
        for vv in 1:4
            layers2 = deepcopy(layers)
            layers2[2].W[k][uu, vv] += epsilon
            _, f2 = forward_propagation(layers2, D, D)
            num_grad_re[k][uu, vv] = (f2 - f) / epsilon
        end
    end
end

println("수치 gradient (실수부) 계산 완료")
```

```julia
# 수치 gradient (허수부): W[k][u,v] → +im*epsilon 섭동
num_grad_im = deepcopy(PT.W)

for k in 1:NW
    for uu in 1:4
        for vv in 1:4
            layers2 = deepcopy(layers)
            layers2[2].W[k][uu, vv] += im * epsilon
            _, f2 = forward_propagation(layers2, D, D)
            num_grad_im[k][uu, vv] = (f2 - f) / epsilon
        end
    end
end

println("수치 gradient (허수부) 계산 완료")
```

```julia
# 해석적 gradient
# backward_propagation은 내부적으로 DeltaRule(PTLayer)도 호출하는데,
# W gradient에는 loss → PT 방향의 delta만 있으면 충분하므로 직접 계산.
γ = Clifford()

# loss layer의 DeltaRule만 직접 계산 (PT 입력으로 들어오는 error)
delta_into_pt = DeltaRule(layers[end], layers[end].activation, u[end], 1)

# PTLayer W에 대한 gradient
d = gradient(layers[2], layers[2].activation, u[2], delta_into_pt, γ)
ana_grad = d[1]  # Vector{Spin_matrix}

println("해석적 gradient 계산 완료")
println("W 크기: ", length(ana_grad), "  (npaths*n_out = ", npaths, "*", n_out, ")")
```

```julia
# 비교: 실수부
println("=== 실수부 비교 (수치 vs 해석) ===")
max_err_re = 0.0
for k in 1:NW
    for uu in 1:4
        for vv in 1:4
            nr = real(num_grad_re[k][uu, vv])
            ar = real(ana_grad[k][uu, vv])
            err = abs(nr - ar)
            if err > max_err_re
                max_err_re = err
            end
        end
    end
end
println("최대 절대오차 (실수부): ", max_err_re)
```

```julia
# 비교: 허수부
println("=== 허수부 비교 (수치 vs 해석) ===")
max_err_im = 0.0
for k in 1:NW
    for uu in 1:4
        for vv in 1:4
            ni = real(num_grad_im[k][uu, vv])  # im 섭동 → 허수 방향 gradient
            ai = imag(ana_grad[k][uu, vv])
            err = abs(ni - ai)
            if err > max_err_im
                max_err_im = err
            end
        end
    end
end
println("최대 절대오차 (허수부): ", max_err_im)
```

```julia
# 첫 번째 weight 행렬 상세 비교
k = 1
println("W[", k, "] 수치 gradient (실수부):")
display(real.(num_grad_re[k].sm))
println("W[", k, "] 해석적 gradient (실수부):")
display(real.(ana_grad[k].sm))
println("차이:")
display(real.(num_grad_re[k].sm) .- real.(ana_grad[k].sm))
```

```julia
# n_out=2 테스트
n_out2   = 2
PT2      = PTLayer(P, U, n_out2)
PT2.W    = 0.1 * PT2.W

loss2    = PtcRayleighQuotient_Layer(D, Identity(), n_out2)
layers2b = [fe, PT2, loss2]

u2b, f2b = forward_propagation(layers2b, D, D)
println("n_out=2 loss = ", f2b)

epsilon2 = 1e-6
NW2 = length(PT2.W)
num_grad2 = deepcopy(PT2.W)

for k in 1:NW2
    for uu in 1:4
        for vv in 1:4
            lay3 = deepcopy(layers2b)
            lay3[2].W[k][uu, vv] += epsilon2
            _, f3 = forward_propagation(lay3, D, D)
            num_grad2[k][uu, vv] = (f3 - f2b) / epsilon2
        end
    end
end

delta2b   = DeltaRule(layers2b[end], layers2b[end].activation, u2b[end], 1)
d2b       = gradient(layers2b[2], layers2b[2].activation, u2b[2], delta2b, γ)
ana_grad2 = d2b[1]

max_err2 = 0.0
for k in 1:NW2
    for uu in 1:4, vv in 1:4
        err = abs(real(num_grad2[k][uu,vv]) - real(ana_grad2[k][uu,vv]))
        max_err2 = max(max_err2, err)
    end
end
println("n_out=2 최대 절대오차 (실수부): ", max_err2)
```
