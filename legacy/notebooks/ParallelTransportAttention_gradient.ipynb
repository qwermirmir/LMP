```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP



```

```julia
NX = 8
NY = 8
NZ = 8
NT = 16
Nwing = 0
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))





delta=1e-8
P=Onestep()
hop=Hopping_operator(U,P)
W=Generate_weight(hop)




```

```julia
layer=PTC_Layer(hop,W)
act=Identity()
act=ModRelu()
Attention=PrallelTransportAttention_Layer(layer,0.1, act ,false,0.1)
layers=[Attention]
y=layers(x)
f=norm2(y-t)
grad1=deepcopy(Attention.Q.W)
grad2=deepcopy(Attention.K.W)
grad3=deepcopy(Attention.V.W)
```

```julia

epsilon=1e-8


for i in 1:length(Attention.Q.W)
    for u in 1:4
        for v in 1:4
            local layer1=deepcopy(layers)
            layer1[1].K.W[i][u,v]+=epsilon
            y1=layer1(x)
            grad1[i][u,v]=(norm2(y1-t)-f)/epsilon
        end
    end
end 

for i in 1:length(Attention.K.W)
    for u in 1:4
        for v in 1:4
            local layer1=deepcopy(layers)
            layer1[1].V.W[i][u,v]+=epsilon
            y1=layer1(x)
            grad2[i][u,v]=(norm2(y1-t)-f)/epsilon
        end
    end
end 
```
