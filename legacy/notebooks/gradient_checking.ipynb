```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
using LinearAlgebra
```

```julia
NX = 2
NY = 2
NZ = 2
NT = 4
Nwing = 0
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
W1=Generate_weight(hop)
W2=Generate_weight(hop)
W3=Generate_weight(hop)
W4=Generate_weight(hop)
```

```julia
layer1=CPTC_Layer(hop,W1)
layer2=PTC_Layer(hop,W2)
act=Softmax()
Norm=NormLayer()
layer1=Attention_Layer(layer1,sqrt(length(x)) , act )
layer2=Attention_Layer(layer2,sqrt(length(x))  ,act,true  )
layer3=PTC_Layer(hop,W3)
layer4=CPTC_Layer(hop,W4)
layers=[Norm,layer2]
```

```julia
y=layer1(x)
```

```julia
f=norm2(layers(y)-t)
grad1=similar(W1)
grad2=similar(W1)
grad3=similar(W1)
```

```julia
epsilon = 1e-6



epsilon=1e-8

for i in 1:9
    for u in 1:4
        for v in 1:4
            local layer=deepcopy(layer1)
            layer.Q.W[i][u,v]+=epsilon
            y1=layer(x)
            grad1[i][u,v]=(norm2(layers(y1)-t)-f)/epsilon
        end
    end
end 

for i in 1:9
    for u in 1:4
        for v in 1:4
            local layer=deepcopy(layer1)
            layer.K.W[i][u,v]+=epsilon
            y1=layer(x)
            grad2[i][u,v]=(norm2(layers(y1)-t)-f)/epsilon
        end
    end
end 

for i in 1:9
    for u in 1:4
        for v in 1:4
            local layer=deepcopy(layer1)
            layer.V.W[i][u,v]+=epsilon
            y1=layer(x)
            grad3[i][u,v]=(norm2(layers(y1)-t)-f)/epsilon
        end
    end
end 

```

```julia
layers = [layer1,Norm,layer2]
γ=Clifford()
u = forward_propagation(layers, x)
delta = backward_propagation(layers, t, u)
d = gradient(layer1, layer1.activation, u[1], delta[3],γ)
```
