```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
using LinearAlgebra
```

```julia
NX = 2
NY = 2
NZ = 2
NT = 4
Nwing = 0
Dim = 4
NC = 3
# const pes =  (2, 2, 2, 2)
# const cuda = true
# const numtrj_in = 2 # the number of operations to take average of performance
# const MDsteps_in = 0 # dummy
# PEs = pes

U = Initialize_Gaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))
```

```julia
params = Dict()
params["Dirac_operator"] = "Wilson"
params["κ"] = 0.141139
# params["faster version"] = true
params["eps_CG"] = 1.0e-8
params["verbose_level"] = 2

D = Dirac_operator(U,x,params)
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
W=Generate_weight(hop)
W1=deepcopy(W)
act=Softmax()
```

```julia
KW=layer1.K.W
QW=layer1.Q.W
VW=layer1.V.W

m=sqrt(length(x))


epsilon = 1e-8
grad = similar(x)  

# 기준값 계산
Kx = Clifford(KW)*hop(y)
Qx = Clifford(QW)*hop(y)
Vx = Clifford(VW)*hop(y)

M = similar(KW)
S = similar(KW)

for i in 1:length(KW)
    M[i] = Matmul!(conj(Qx[i]), Kx[i])/m
    S[i] = act(M[i])
end




u = similar(t)  # similar가 0 초기화라고 가정
for i in 1:length(KW)
    u+=S[i]*Vx[i]
end


f = norm2(u-t)
```

```julia
for ic = 1:x.NC, it = 1:x.NT, iz = 1:x.NZ, iy = 1:x.NY, ix = 1:x.NX, fg = 1:x.NG

    d = deepcopy(y)
    d[ic, ix, iy, iz, it, fg] += epsilon

    Kd = Clifford(KW)*hop(d)
    Qd = Clifford(QW)*hop(d)
    Vd = Clifford(VW)*hop(d)

    dM = similar(KW)
    dS = similar(KW)

    for i in 1:length(KW)
        dM[i] = Matmul!(conj(Qd[i]), Kd[i])/m
        dS[i] = act(dM[i])
        # dS[i] = dM[i]
    end



    k = similar(t)  # similar가 0 초기화라고 가정
    for i in 1:length(KW)
        k += dS[i] * Vd[i]
    end




    grad[ic, ix, iy, iz, it, fg] = (norm2(k-t) - f) / epsilon
end

```

```julia
delta = similar(x)  # 이미 0 초기화

# 정확한 gradient 방향 (손실이 norm^2이므로)
d = similar(x)

# eta 정확한 계산 (필수!)
eta = similar(KW)
bareta = similar(KW)

for i in 1:length(KW)
    eta[i] = Matmul!(Vx[i], conj(u))

end

for i in 1:length(KW)
    
    bareta[i]= Matmul!(conj(Vx[i]), u)

end
# bareta= Matmul!(y, conj(t))


# Kz 계산 (정확한 버전)
Kz1 = similar(KW)

for i in 1:length(KW)
    for a in 1:4, c in 1:4 , m in 1:4
        Kz1[i][a,c] +=Clifford(KW)[i][m,a]* S[i][c,m]*eta[i][m,c]
    end
end

Kz2 = similar(KW)

for i in 1:length(KW)
    for a in 1:4, d in 1:4, c in 1:4 , m in 1:4
        Kz2[i][a,c] +=Clifford(KW)[i][d,a]*S[i][c,m]*S[i][c,d]*eta[i][m,c]
    end
end
Kz = Kz1 - Kz2




# # 최종 delta 계산
for i in 1:length(KW)
    delta += Kz[i]*conj(Clifford(QW)[i]*y)/m
end




Qz1 = similar(QW)
Qz2 = similar(QW)


for i in 1:length(QW)
    for a in 1:4,  c in 1:4, d in 1:4
        Qz1[i][a,c] +=Clifford(QW)[i][d,a]*(adjoint(S[i])[c,d])*bareta[i][c,d]
    end
end

for i in 1:length(QW)
    for a in 1:4, c in 1:4, e in 1:4 , d in 1:4
        Qz2[i][a,c] +=  Clifford(QW)[i][d,a]*(adjoint(S[i])[e,d])*(adjoint(S[i])[c,d]) * bareta[i][e,d]
    end
end







Qz=Qz1-Qz2

for i in 1:length(QW)
    delta+=Qz[i]*conj(Clifford(KW)[i]*y)/m

end




invP = reverse_paths(P)
invhop = Hopping_operator(U, invP)

Vz=similar(VW)

for i in 1:length(VW)
    for a in 1:4
        for c in 1:4
            for e in 1:4
                Vz[i][a,c]+=Clifford(VW)[i][e,a]*S[i][c,e]
            end
        end
    end
end


dH=invhop(u)

for i in 1:length(VW)
    delta+=Vz[i]*conj(dH[i])
end    


delta =2*delta
```
