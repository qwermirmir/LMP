```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 0
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
t = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(t)
x=x/sqrt(norm2(x))
t=t/sqrt(norm2(t))
gamma=Clifford()
```

```julia
P=Paths([1])
hop=Hopping_operator(U,P)
W=Generate_weight(hop)
```

```julia
layer=CPTC_Layer(hop,W)
```

```julia
y=norm2(layer(x)-t)
delta=1e-9
```

```julia
delta = 1e-8  # 부동소수점 오차를 방지하기 위한 delta 값 조정

# Gradient 저장할 행렬
grad1 = deepcopy(W)
L1 = deepcopy(W)

# f 계산 (원래 값)
y_ij = similar(x)
for v in 1:2
    for i in 1:4
        for j in 1:4
            y_ij += W[v][i, j] * gamma[4 * (i-1) + j] *hop(x)[v]
        end
    end
end 
f = norm2(y_ij-t)

# # 수치 미분을 통한 gradient 계산
for v in 1:2
    for i in 1:4
        for j in 1:4
            # W perturbation 적용
            dW_plus = deepcopy(W)
            dW_minus = deepcopy(W)
            
            dW_plus[v][i, j] += delta
            dW_minus[v][i, j] -= delta

            # Perturbation을 적용한 y_ij 계산
            y_ij_plus = similar(x)
            y_ij_minus = similar(x)
            for k in 1:2
                for a in 1:4
                    for b in 1:4
                        y_ij_plus += dW_plus[k][a, b] * gamma[4 * (a-1) + b] * hop(x)[k]
                        y_ij_minus += dW_minus[k][a, b] * gamma[4 * (a-1) + b] *hop(x)[k]
                    end
                end
            end 
            f_plus = norm2(y_ij_plus-t)
            f_minus = norm2(y_ij_minus-t)

            # 중앙 차분 방식으로 수치 미분
            grad1[v][i, j] = (f_plus - f_minus) / (2 * delta)
        end
    end
end 

# 해석적 미분 계산
for v in 1:2
    for i in 1:4
        for j in 1:4
            L1[v][i, j] = inner_product(y_ij-t, gamma[4 * (i-1) + j] *hop(x)[v])
        end
    end
end 

# Hermitian 대칭성 적용
L1 = conj(L1) + L1

# 수치 미분과 해석적 미분 비교
println("Gradient Difference (L - grad): ", L1 - grad1)
```
