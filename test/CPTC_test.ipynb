```julia
using LatticeDiracOperators
using LatticeQCD
using Gaugefields
using Wilsonloop
using LMP
```

```julia
NX = 4
NY = 4
NZ = 4
NT = 4
Nwing = 1
Dim = 4
NC = 3
U = Initialize_4DGaugefields(NC,Nwing,NX,NY,NZ,NT,condition = "hot")
x = Initialize_pseudofermion_fields(U[1],"Wilson",nowing=true)
gauss_distribution_fermion!(x)
```

```julia
P=Onestep()
hop=Hopping_operator(U,P)
```

```julia
W=Generate_weight(hop)
W2=[W[1]]
```

```julia
layer1 = CPTC_Layer(hop, W2)
```

```julia
sequence=[layer1]
```

```julia
f,u=forward_propagation(sequence,x)
```

```julia
println(f[1,1,1,1,1,1])
```

```julia
rP=reverse_paths(P)
```

```julia
rhop=Hopping_operator(U,rP)
```

```julia
layer2 = CPTC_Layer(rhop, W2)
```

```julia
A=layer1(x)
B=layer2(x)
```

```julia
println(A[2,1,1,1,1,1])
println(B[2,1,1,1,1,1])
```

```julia
norm2(A-B)
```
